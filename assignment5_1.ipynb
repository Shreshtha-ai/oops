```python
class Employee:

    def __init__(self):
        self.name = "Ravi"             # Public
        self._department = "CSE"       # Protected
        self.__salary = 50000          # Private


# Create object
emp = Employee()

# Public variable
emp.name = "Amit"

# Protected variable
emp._department = "ECE"

emp.__salary = 70000

# Print all three values
print("Name:", emp.name)
print("Department:", emp._department)
print("Salary using emp.__salary:", emp.__salary)

print("Original Salary:", emp._Employee__salary)
```

```python
class Patient:

    def __init__(self):
        self.name = "Rahul"             # Public
        self._age = 30                  # Protected
        self.__disease = "Fever"        # Private


# Create object
patient = Patient()

# Public variable
print("Patient Name:", patient.name)

# Protected variable
print("Patient Age:", patient._age)

print("Patient Disease:", patient._Patient__disease)

try:
    print(patient.__disease)
except AttributeError:
    print("Error: Cannot access __disease directly")
```

```python
class DigitalWallet:

    def __init__(self, owner_name, security_code):
        self.owner_name = owner_name          # Public
        self._wallet_type = "Digital Wallet"  # Protected
        self.__balance = 0                    # Private
        self.__security_code = security_code  # Private

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount, security_code):

        if security_code != self.__security_code:
            print("Incorrect security code.")
            return

        if amount > self.__balance:
            print("Insufficient balance.")
            return

        if amount <= 0:
            print("Invalid withdrawal amount.")
            return

        self.__balance -= amount
        print("Amount withdrawn:", amount)

    def check_balance(self):
        print("Current balance:", self.__balance)

    def change_security_code(self, old_code, new_code):

        if old_code == self.__security_code:
            self.__security_code = new_code
            print("Security code changed successfully.")
        else:
            print("Incorrect old security code.")


# Main program

wallet = DigitalWallet("Amit", 1234)

wallet.deposit(5000)

wallet.check_balance()

wallet.withdraw(1000, 1234)

wallet.check_balance()

wallet.withdraw(10000, 1234)

wallet.change_security_code(1234, 5678)

wallet.withdraw(500, 5678)

wallet.check_balance()
```
