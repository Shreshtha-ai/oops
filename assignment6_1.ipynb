```python
#CASE-3
class UniversityMember:

    def __init__(self,name,member_id):
        self.name = name
        self.member_id = member_id
class Student(UniversityMember):
    def __init__(self,name,member_id,course, semester):
        super().__init__(name,member_id)
        self.course = course
        self.semester = semester
class PostgraduateStudent(Student):
    def __init__(self,name,member_id,course,semester,specialization,research_area):
        super().__init__(name,member_id,course,semester)
        self.specialization = specialization
        self.research_area = research_area

    def display_profile(self):
        print("Name:", self.name)
        print("Member ID:", self.member_id)
        print("Course:", self.course)
        print("Semester:", self.semester)
        print("Specialization:", self.specialization)
        print("Research Area:", self.research_area)


student1 = PostgraduateStudent("abc", "1234", "Computer Science", 2, "Artificial Intelligence", "Machine Learning")
student2 = PostgraduateStudent("xyz","56788", "Data Science", 1, "Big Data", "Data Analytics")

student1.display_profile()
student2.display_profile()
```

```python
#CASE-5

class BankAccount:
    def __init__(self,account_number,holder_name,balance):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self,amount):
        self.balance +=amount

    def display_balance(self):
        print("balance :",self.balance)

class SavingsAccount(BankAccount):
    def __init__(self,account_number,holder_name,balance,interest_rate):
        super().__init__(account_number,holder_name,balance)
        self.interest_rate = interest_rate
    def calculate_interest(self):
        interest = self.balance * (self.interest_rate / 100)
        return interest

class CurrentAccount(BankAccount):
    def __init__(self,account_number,holder_name,balance,overdraft_limit):
        super().__init__(account_number,holder_name,balance)
        self.overdraft_limit = overdraft_limit
    def check_overdraftt(self):
        if self.balance >=self.overdraft_limit:
            print("Overdraft is allowed")

        else:
            print("Not sufficeient balance")

savings = SavingsAccount(101, "abc", 890, 2)

current = CurrentAccount(102, "xyz", 1000,5000)

print("Savings Account:")
savings.display_balance()

print("Intreset:",savings.calculate_interest())

print("Current aCCOUNT")

current.display_balance();

current.check_overdraftt()

```
