```python
class Election:
    def __init__(self,state,voters =None,leader=None): #leader and voter are optional argument
        self.state = state
        self.voters = voters
        self.leader = leader

    def dispaly(self):
        print("State:",self.state)
        print("Voters:",self.voters)
        print("Leaders:",self.leader)

punjab = Election("punjab",200000)
up = Election("UP",3000000)
bjp  = Election("Bihar",20000,"Aman")

print("Voters of punjab:",punjab.voters)
print("Voters of up:",up.voters)
print("Leader of Bihar:",bjp.leader)

bjp.leader = "abc"
print("New Leader of bihar:",bjp.leader)

punjab.dispaly()


```

```python
class SmartAppliance:
    def __init__(self,appliance_name,room_location,power_rating_watts,is_on = False):
        self.appliance_name = appliance_name
        self.room_location = room_location
        self.power_rating_watts = power_rating_watts
        self.is_on =is_on

    def turn_on(self):
        self.is_on = True
        print(self.appliance_name, "is now ON")
    def turn_off(self):
        self.is_on = False
        print(self.appliance_name, "is off")

    def display(self):
        print("Appliance Name:", self.appliance_name)
        print("Room Location:", self.room_location)
        print("Power Rating:", self.power_rating_watts, "watts")

        if self.is_on == True:
            print("Status: ON")
        else:
            print("Status: OFF")
ac_unit = SmartAppliance("Air Conditioner", "Living Room", 1500)

smart_light = SmartAppliance("Smart Light", "Bedroom", 20)



ac_unit.display()
ac_unit.turn_on()

smart_light.display()
smart_light.turn_on()



```

```python
class EnergySubstation:
    city_name = "Neo Metro"
    total_city_power_drawn = 0.0

    def __init__(self, substation_id, max_capacity_kw):
        self.substation_id = substation_id
        self.max_capacity_kw = max_capacity_kw
        self.current_load_kw = 0.0

    def add_load(self, amount):
        if self.current_load_kw + amount <= self.max_capacity_kw:
            self.current_load_kw += amount

            EnergySubstation.total_city_power_drawn += amount

            print(f"{self.substation_id}: Added {amount} kW load.")
            print(f"Current load: {self.current_load_kw}/{self.max_capacity_kw} kW")
        else:
            print(f"{self.substation_id}: Overload warning Cannot add {amount} kW load.")

    def display_status(self):
        print("Substation ID:", self.substation_id)
        print("Current Load:", self.current_load_kw, "kW")
        print("Maximum Capacity:", self.max_capacity_kw, "kW")
        print("City:", EnergySubstation.city_name)


sub_1 = EnergySubstation("SUB_NORTH", 500.0)
sub_2 = EnergySubstation("SUB_SOUTH", 300.0)

sub_1.add_load(350.0)

sub_2.add_load(200.0)

sub_2.add_load(150.0)  

sub_2.max_capacity_kw = 400.0

sub_2.add_load(100.0)

print("Total city power drawn:", EnergySubstation.total_city_power_drawn, "kW")




```

```python
class Player: 

    def __init__(self, username): 
        self.username = username 
        self.__hp = 100 

 

    def take_damage(self, amount): 
        self.__hp = self.__hp - amount 
        if self.__hp < 0: 
            self.__hp = 0 
        print(self.username, "took", amount, "damage, hp:", self.__hp) 

 

    def heal(self, amount): 

        self.__hp = self.__hp + amount 

        if self.__hp > 100: 

            self.__hp = 100 

        print(self.username, "healed", amount, "hp:", self.__hp) 

 

    def is_alive(self): 

        return self.__hp > 0 

 

p1 = Player("abc") 

p1.take_damage(40) 

p1.heal(10) 

print("Alive:", p1.is_alive()) 

 

p1.__hp = 9999 

print(p1.__hp) 

print(p1._Player__hp) 
```
