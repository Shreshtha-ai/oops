```python
class BankAccount:
    def __init__(self, account_number, account_name,balance):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance
    def deposit(self, amount):
        self.balance = self.balance+ amount
        print("Amount deposited:", amount)
    def withdraw(self, amount ):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Amount withdrawn", amount)

        else:
            print("Not enough balance")
    def display(self):
         print("Account Number:", self.account_number)
         print("Account Name:", self.account_name)
         print("Current balance:",self.balance)

account_number = int(input("Enter Account Number: "))
account_name = input("Enter Account Name: ")

balance = float(input("Enter Initial Balance: "))
b = BankAccount(account_number, account_name , balance)
amount = float(input("Enter Amount to Deposit: "))
b.deposit(amount)

amount = float(input("Enter the amount to be withdrawn:"))
b.withdraw(amount)

b.display()






```

```python
class LibraryBook:
    def __init__(self, book_name, author, copies):
        self.book_name = book_name
        self.author = author
        self.copies = copies
    def issue_book(self):
        if self.copies > 0:
            self.copies = self.copies - 1
            print("Book Issued")
        else:
            print("Book Not Available")
    def return_book(self):
        self.copies = self.copies + 1
        print("Book Returned")


    def display(self):
        print("Book name:", self.book_name)
        print("Author:", self.author)
        print("Number of copies:", self.copies)


book_name = input("Enter Book name: ")
author = input("Enter author: ")
copies = int(input("Enter number of copies: "))

b = LibraryBook(book_name, author, copies)

b.display()

b.issue_book()
b.display()

b.return_book()
b.display()
```

```python
class Calculator:

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        print("Addition:", self.num1 + self.num2)

    def subtract(self):
        print("Subtraction:", self.num1 - self.num2)

    def multiply(self):
        print("Multiplication:", self.num1 * self.num2)

    def divide(self):
        if self.num2 != 0:
            print("Division:", self.num1 / self.num2)
        else:
            print("Cannot divide by zero")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

c = Calculator(num1, num2)


c.add()
c.subtract()
c.multiply()
c.divide()
```

```python
class Car:

    def __init__(self, brand, model, year, price):
        self.brand = brand
        self.model = model
        self.year = year
        self.price = price

    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Year:", self.year)
        print("Price:", self.price)

    def update_price(self, new_price):
        self.price = new_price
        print("Price Updated")


brand = str(input("Enter brand: "))
model = str(input("Enter model: "))
year = int(input("Enter year: "))
price = float(input("Enter price: "))

c = Car(brand, model, year, price)

c.display()

new_price = float(input("Enter New Price: "))

c.update_price(new_price)

c.display()
```

```python
class Mobilephone:

    def __init__(self, brand, model, storage, battery_percentage):
        self.brand = brand
        self.model = model
        self.storage = storage
        self.battery_percentage = battery_percentage

    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Storage:", self.storage)
        print("Battery Percentage:", self.battery_percentage)

    def charge(self, amount):
        self.battery_percentage = self.battery_percentage + amount

        if self.battery_percentage > 100:
            self.battery_percentage = 100

        print(" Charged")


    def battery_status(self):
        if self.battery_percentage <= 30:
            print("Low Battery")
        else:
            print("Battery Level is Good")


brand = input("Enter brand: ")
model = input("Enter model: ")
storage = input("Enter storage: ")
battery = int(input("Enter battery percentage: "))

m = Mobilephone(brand, model, storage, battery)

m.display()
m.battery_status()

amount = int(input("Enter charging percentage: "))

m.charge(amount)

m.display()
m.battery_status()
```

```python
class Product:

    def __init__(self, product_name, product_id, price, quantity):
        self.product_name = product_name
        self.product_id = product_id
        self.price = price
        self.quantity = quantity

    def total_bill(self):
        total = self.price * self.quantity
        print("Total Bill:", total)

    def display(self):
        print("Product name:", self.product_name)
        print("Product ID:", self.product_id)
        print("Price:", self.price)
        print("Quantity:", self.quantity)


product_name = str(input("Enter Product Name: "))
product_id = int(input("Enter Product ID: "))
price = float(input("Enter Price: "))
quantity = int(input("Enter Quantity: "))

p = Product(product_name, product_id, price, quantity)

p.display()
p.total_bill()
```

```python
class Bookstore:


    def __init__(self, book_name, author, price, copies):
        self.book_name = book_name
        self.author = author
        self.price = price
        self.copies = copies

    def display(self):
        print("Book name:", self.book_name)
        print("Author:", self.author)
        print("Price:", self.price)
        print("Number of copies:", self.copies)

    def sell_book(self):
        if self.copies > 0:
            self.copies = self.copies - 1
            print("Book Sold")
        else:
            print("Book Not Available")

    def add_copies(self, number):
        self.copies = self.copies + number
        print("Copies Added")


book_name = input("Enter book : ")
author = input("Enter author: ")
price = float(input("Enter price: "))
copies = int(input("Enter number of copies: "))

b = Bookstore(book_name, author, price, copies)

b.display()

b.sell_book()
b.display()
number = int(input("Enter Number of New Copies: "))

b.add_copies(number)

b.display()
```

```python
class Attendance:

    def __init__(self, student_name, total_classes, attended_classes):
        self.student_name = student_name
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    def attendance_percentage(self):
        percentage = (self.attended_classes / self.total_classes) * 100

        print("Student name:", self.student_name)
        print("Total classes:", self.total_classes)
        print("Attended classes:", self.attended_classes)
        print("Attendance percentage:", percentage)

    def check_eligibility(self):
        percentage = (self.attended_classes /self.total_classes) * 100

        if percentage >= 75:
            print("Student is eligible for examination")
        else:
            print("Student is not eligible for examination")


student_name = input("Enter student name: ")
total_classes = int(input("Enter total classes: "))
attended_classes = int(input("Enter attended classes: "))

a = Attendance(student_name, total_classes, attended_classes)

a.attendance_percentage()
a.check_eligibility()
```
