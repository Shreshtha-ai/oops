```python
#Case-1
class BankAccount:
    
    def __init__(self, owner, initial_balance=0):
        self.__owner = owner
        self.__initial_balance = initial_balance
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f" Deposited ${amount}")
        else:
            print("Deposit amount must be positive.")
    
    def withdraw(self, amount):
        if amount>0 and amount<= self.__balance:
            self.__balance -= amount
            print(f" Withdrew ${amount}")
        else:
            print("Invalid withdrawal amount or insufficient funds.")
    def get_balance(self):
        return self.__balance
    
account1 = BankAccount("abc",100000)
print(account1.get_balance())
account1.__balance = 50000
print(account1.get_balance())
```

```python
class Order: 

    def __init__(self): 

        self._shipping_status = "Processing" 

 

class ExpressOrder(Order): 

    def expedite_shipping(self): 

        self._shipping_status = "In Transit - Priority" 

 

    def track_order(self): 

        return self._shipping_status 

 

order1 = ExpressOrder() 

print(order1.track_order()) 

order1.expedite_shipping() 

print(order1.track_order())
```
